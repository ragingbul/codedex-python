```python
import random
choices = ('✊', '✋', '✌️')

player_choice = input("Enter ✊, ✋, or ✌️: ").lower()
computer_choice = random.choice(choices)

if player_choice == computer_choice:
    print("tie")
elif (player_choice == "✊" and computer_choice == "✌️") or \
     (player_choice == "✌️" and computer_choice == "✋") or \
     (player_choice == "✋" and computer_choice == "✊"):
    print("you win !!!!")
else:
    print("you loose")
```
