```python
# dry.py

# --- 4 built-in functions we've already used before ---

# 1. print() -> Displays output to the console
print("Hello, world!")

# 2. input() -> Takes user input as a string
name = input("Enter your name: ")

# 3. len() -> Returns the length of an object (like a string or list)
print("Your name has", len(name), "characters.")

# 4. int() -> Converts a string or float into an integer
age = int(input("Enter your age: "))
print("In 5 years, you will be", age + 5)


# --- 1 new built-in function we haven’t used before ---

# 5. sum() -> Takes an iterable (like a list or tuple) and returns the total sum of its elements
numbers = [10, 20, 30, 40, 50]
print("The total sum of the list is:", sum(numbers))




```
